1.Triangle area and perimeter

```python
import math
a=3
b=4
c=5
perimeter=a+b+c
s=perimeter/2
area=math.sqrt(s*(s-a)*(s-b)*(s-c))
print("Area= ",area,"Perimeter= ",perimeter)
```

```python
#for circle
import math
r=1
area=math.pi*r**2
circumference=2*math.pi*r
print("Area= ",area,"Circumference= ",circumference)
```

2.Basis Dimesion (Linear algebra)



```python
import numpy as np
A=np.array([
    [1,2,3],
    [0,1,4],
    [5,6,0]
])

#determinant and inverse
print("Determinant= ",np.linalg.det(A))
print("Inverse of A= ",np.linalg.inv(A))
```

```python
import sympy as sp
A= A = sp.Matrix([
    [1,2,3],
    [2,4,6],
    [1,1,1]
])
print("Basis= ",A.columnspace())
print("Dimension= ",A.rank())
print("Nullity= ",A.nullspace())
```

3.Inequality solve x^2-5x+6>0

```python
import sympy as sp
x=sp.symbols('x')
ineq=x**2-5*x+6>0
solution=sp.solve_univariate_inequality(ineq,x)
print(solution)
```

4.Quadratic equation solve

```python
import sympy as sp
x=sp.symbols('x')
eqn=sp.Eq(x**2-5*x+1,0)
roots=sp.solve(eqn,x)
for y in roots:
  print(y)
```

5. Pair of straight lines

```python

import matplotlib.pyplot as plt
import numpy as np
import math
a=1
two_h=-3
b=2
d=(two_h)**2-(4*a*b)
x=np.linspace(-10,10,100)
if d<0:
  print("The two lines are imaginary and cannot be plotted.")
elif d==0:
 print("The two lines are real and coincident.")
 m=-two_h/(2*b)
 y=m*x
 plt.plot(x,y,label=f"y={m}x")
 plt.xlabel('x axis')
 plt.ylabel('y axis')
 plt.title("Pair of straight lines")
 plt.grid(True)
 plt.show()
else:
  print("The two lines are real and distinct")
  m1=-two_h/(2*b)+math.sqrt(d)/(2*b)
  m2=-two_h/(2*b)-math.sqrt(d)/(2*b)
  y1=m1*x
  y2=m2*x
  plt.plot(x,y1,label=f"y={m1}x")
  plt.plot(x,y2,label=f"y={m2}x")
  plt.title("Pair of straight lines")
  plt.xlabel("x axis")
  plt.ylabel("y axis")
  plt.grid(True)
  plt.legend()
  plt.show()
```

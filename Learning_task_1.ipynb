# Scikit-learn

```python
from sklearn.linear_model import LinearRegression
import pandas as pd
student_data={
    "name":["monjil","sami","shovon","rafi","parvez"],
    "hours":[2.5,4,1.5,5.5,3],
    "marks":[65,78,52,92,70]
}
df_student=pd.DataFrame(student_data)
x=df_student[["hours"]]
y=df_student["marks"]
model=LinearRegression()
model.fit(x,y)
z=5.9
predict=model.predict([[z]])
print("Prediction for z = ",z,"is",predict[0])

```
